### Question 3: What is the distribution of Type II respiratory failure?
 Reasoning: Analyzing the distribution of Type II respiratory failure helps determine how common this condition is among heart failure patients and provides insight into the respiratory health profile of the study population.

```python
df['type_ii_respiratory_failure'].value_counts()
```

```python
df['type_ii_respiratory_failure'].value_counts().plot(
    kind='bar',
    figsize=(6, 4)
)

plt.xlabel('Respiratory Failure Type')
plt.ylabel('Number of Patients')
plt.title('Distribution of Type II Respiratory Failure')
plt.xticks(rotation=0)
plt.show()
```

### Key Insight:
The majority of heart-failure patients did **not** have Type II respiratory failure.Out of **2,008 patients**, **1,894 (94.3%)** were classified as non-Type II respiratory failure. Only **114 patients (5.7%)** were classified as having Type II respiratory failure. This indicates that **Type II respiratory failure is relatively uncommon** in this study population.
