0

```python
binary_cols_with_missing = [
    'peptic_ulcer_disease',
    'moderate_to_severe_chronic_kidney_disease',
    'liver_disease'
]

df[binary_cols_with_missing] = (
    df[binary_cols_with_missing].astype('Int64')
)
```
